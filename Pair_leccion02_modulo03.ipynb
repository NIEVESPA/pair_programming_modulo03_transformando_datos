 A partir de las dos listas anteriores crea un diccionario usando un bucle for. Para hacer esto tenemos el método zip que nos permite iterar por dos listas a la vez. Más información sobre este método aquí.


```python
valores = [1, 5, 7, 2, 4, 3, 6]

indices = ["Perro", "Gato", "Pájaro", "Pez", "Conejo", "Tortuga", "Hamster"]

combinacion = zip(valores, indices) ## las combinamos con el metode zip
combinacion
```

```python
lista_combinada = list(combinacion) ## rompemos la estrucura zip transformando a lista de tuplas
lista_combinada
```

se puede hacer: lista_combinada = list(zip(valores, indices))

```python
lista_dict = dict(zip(valores, indices)) ## sin bucle for
lista_dict
```

```python

dict_lista = {}

for clave, valor in zip(valores,indices):
    dict_lista[clave] = valor
print(dict_lista)
```

Crea una Serie con el diccionario creado en el punto anterior.

```python
serie_diccionario = pd.Series(dict_lista)
serie_diccionario
```
